```python
import random
suits = ["Clubs", "Spades", "Hearts", "Diamonds"]
faces = ["Jack", "Queen", "King", "Ace"]
numbered = [2, 3, 4, 5, 6, 7, 8, 9, 10]
```

```python
for suit in suits:
    for card in faces + numbered:
        deck.add((card, "of", suit))
```

```python
def draw ():
    card = random.choice(list(deck))
    deck.remove(card)
    return card
len(deck)
```
