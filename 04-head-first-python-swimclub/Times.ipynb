```python
#1/ Extract data from the file’s name
#2/ Process the data in the file.

#first we select file and folder then removes redundant parts and split into clean meaningful parts from names, catergories. then we read times data, determine which file lines to read and and strip and split to get "times" list. Then create a "convers" list [] then loop thru times split mins, secs, hundredths and calculate into hundredths format and append to "converts" list then from converts [] list, use import statistics to get mean value, and then convert back to x:x.xx format for average value finally print out the "swimer, age, distance, stroke,"" "times" and "average value".
```

```python
FN = "Darius-13-100m-Fly.txt"
FOLDER = "swimdata/"
swimer, age, distance, stroke = FN.removesuffix(".txt").split("-")
print(swimer)
print(age)
print(distance) 
print(stroke)
```

```python
with open(FOLDER + FN) as file:
    lines = file.readlines()
    times = lines[0].strip().split(",") #strip() removes "spaces" include /n and then split at the "," to get a list of times.
```

```python
converts = []
for t in times:
    minutes, rest = t.split(":")
    seconds, hundredths = rest.split(".")
    converts.append((int(minutes) * 60 * 100) + (int(seconds) * 100) + int(hundredths))
```

```python
import statistics

average = statistics.mean(converts) #call mean for converts list get average value

mins_secs, hundredths = str(round(average / 100, 2)).split(".") #divide by 100 to get the minutes and seconds value from hundredths form. round to 2 decimal places and then split at the "." to get the hundredths value.

mins_secs = int(mins_secs) #int for calculation 
minutes = mins_secs // 60
seconds = mins_secs - minutes *60 #minutes is 1 so "*60" for value in secs.
average = str(minutes) + ":" + str(seconds) + "." + hundredths #format back again 
```
