```python
#AI code to fix HTTP Error 403: Forbidden,a workaround for the error with using gazpacho to change User-agent identity to a normal web brower
import pandas as pd
import gazpacho
from io import StringIO

URL = "https://en.wikipedia.org/wiki/List_of_world_records_in_swimming"

html = gazpacho.get(URL)
tables = pd.read_html(StringIO(html))

len(tables)
```

```python
RECORDS = (0, 1, 3, 4)
COURSES = ("LC Men", "LC Women", "SC Men", "SC Women")

records = {}
for table, course in zip(RECORDS, COURSES):
    df = tables[table][["Event", "Time"]] #Inner [] creates a list of 2 column names. Outer [] tells Pandas to select those columns.
    df = df[~df["Event"].str.contains("relay")] #exclude relay records
    df = df.set_index("Event") #set the index to the Event column
    records[course] = df.to_dict()["Time"] #get the dictionary of times from the dataframe and assign it to the records dictionary under the key "course"
```

```python
import json

with open("records.json") as jf:
    gazpacho_records = json.load(jf)

records == gazpacho_records
```

```python
for course in records:
    print(course, records[course] == gazpacho_records[course])
```
