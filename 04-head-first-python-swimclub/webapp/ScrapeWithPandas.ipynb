```python
#AI code to fix HTTP Error 403: Forbidden,a workaround for the error with using gazpacho to change User-agent identity to a normal web brower
import pandas as pd
import gazpacho
from io import StringIO

URL = "https://en.wikipedia.org/wiki/List_of_world_records_in_swimming"

html = gazpacho.get(URL)
tables = pd.read_html(StringIO(html))

len(tables)
```
