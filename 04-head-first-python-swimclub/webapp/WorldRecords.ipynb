```python
for n, table in enumerate(tables):
    rows = table.find("tr", mode="all")
    cols=rows[-1].find("td", mode="all")
    print(f"{n} -> {len(rows)} rows, {len(cols)} columns")
    #tr = table row return list and cols = -1 row from below and find td ( table data cells) count td horizontally then print and we will get 0, 1, 3, 4 as main output to process next
```

```python
table = tables[0]
for row in table.find("tr", mode="all")[1:]:  # Skip the header row
    columns = row.find("td", mode="all")
    event = columns[0].text 
    time = columns[1].text
    print(f"{event} -> {time}")
```

```python
RECORDS = (0, 1, 3, 4)
COURSES = ("LC Men", "LC Women", "SC Men", "SC Women") 
```

```python
RECORDS = (0, 1, 3, 4)
COURSES = ("LC Men", "LC Women", "SC Men", "SC Women")

for table, course in zip(RECORDS, COURSES):
    print(f"{course}:")
    
    for row in tables[table].find("tr", mode="all")[1:]:
        columns = row.find("td", mode="all")
        event = columns[0].text
        time = columns[1].text
        print(f"\t{event} -> {time}")
    
    print()
```

```python
records = {} #create an empty dictionary to store the records for each course in next step pair

for table, course in zip(RECORDS, COURSES): #zip paired, course name from 0->last and its table 0->last (from RECORDS listed) of each course    
    records[course] = {} #eg: records["LC Men"] = {} 0->last, each time change

    for row in tables[table].find("tr", mode="all")[1:]: #nested loop to iterate through each row of the table for the current course, skipping the header row and save event and time from columns 0 and 1
        columns = row.find("td", mode="all")
        event = columns[0].text
        time = columns[1].text
        records[course][event] = time #eg: records["LC Men"]["100m freestyle"] = "46.86"
```
