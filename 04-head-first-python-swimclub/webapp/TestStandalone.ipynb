```python
def event_lookup(event):
    conversions = {
        "Free": "freestyle",
        "Back": "backstroke",
        "Breast": "breaststroke",
        "Fly": "butterfly",
        "IM": "individual medley",
    }
    *_ , distance, stroke = event.removesuffix(".txt").split("-")

    return f"{distance} {conversions[stroke]}"
```

```python
event_lookup("Mike-15-100m-Fly.txt")
```
