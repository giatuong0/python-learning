```python
import os
import swimclub

swim_files = os.listdir(swimclub.FOLDER)
swim_files.remove('.DS_Store')  # Remove the macOS metadata file

print(len(swim_files)) 
```

```python

swimmers = []
for file in swim_files:
    swimmers.append(swimclub.read_swim_data(file)[0])
print(swimmers)
```

```python
names = list(set(swimmers))  # Remove duplicates by converting to a set, then back to a list
```

```python
print(names)
```

```python
len(names)
```

```python
swimmers = []
for file in swim_files:
    name = swimclub.read_swim_data(file)[0]
    if name not in swimmers:
        swimmers.append(name)
print(swimmers)
```

```python
print(sorted(names))  # Sort the names list set list
```

```python
print((sorted(swimmers)))  # Sort the names on fly for loop
```

```python
swimmers={}
for file in swim_files:
    name, *_ = swimclub.read_swim_data(file)
    if name not in swimmers:
        swimmers[name] = []
    swimmers[name].append(file)
```
